```python
#2
import numpy as np
import scipy.stats as stats

amostra = np.array([48, 53, 42, 67, 24, 54, 61, 73, 18, 52])
n = len(amostra)
media = np.mean(amostra)
dp = np.std(amostra, ddof=1)

t_crit = stats.t.ppf(0.975, df=n - 1)
erro = t_crit * (dp / np.sqrt(n))

ic_inf = media - erro
ic_sup = media + erro

print(f"Média: {media:.2f}")
print(f"Intervalo de Confiança (95%): [{ic_inf:.2f}, {ic_sup:.2f}]")
```

```python
#4
import numpy as np
import scipy.stats as stats

dados = np.array(
    [405, 348, 365, 291, 135, 260, 300, 155, 34, 294, 758, 472, 559, 143, 172]
)
n = len(dados)
media = np.mean(dados)
dp = np.std(dados, ddof=1)

# a) Intervalo de Confiança 95%
t_crit = stats.t.ppf(0.975, df=n - 1)
erro = t_crit * (dp / np.sqrt(n))

ic_inf = media - erro
ic_sup = media + erro

# b) Número esperado de intervalos fora da média
grupos = 40
nao_contem = grupos * 0.05

print(f"a) Média: {media:.2f} | IC (95%): [{ic_inf:.2f}, {ic_sup:.2f}]")
print(f"b) Espera-se {int(nao_contem)} intervalos.")
```
