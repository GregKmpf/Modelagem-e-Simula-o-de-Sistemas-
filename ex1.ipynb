```python
#1
import scipy.stats as stats
import numpy as np

s1 = np.array([13, 12, 16, 14, 15, 10, 13, 11])
s2 = np.array([14, 17, 14, 16, 15, 17, 12, 19])

t_stat, p_val = stats.ttest_ind(s1, s2)

print(f"t-estatística: {t_stat:.4f}")
print(f"p-valor: {p_val:.4f}")
if p_val > 0.05:
  print("O modelo É VÁLIDO com 95% de confiança.")
else:
  print("O modelo NÃO É VÁLIDO com 95% de confiança.")
```
